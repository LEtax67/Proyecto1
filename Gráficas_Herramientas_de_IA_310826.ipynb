```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,5,10,) #inicio, fin, cantidad de datos
y = x**2

plt.plot(x,y)
plt.show() # quita el encabezado de la gráfica
```

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5,5,10,) #inicio, fin, cantidad de datos
y = x**2

plt.plot(x,y, color = "red", marker = "o", linestyle = ":", linewidth = "1")
plt.title("$f(x)=x^2$")
plt.xlabel("$x$")
plt.ylabel("$x^2$")
plt.grid() # agrega una grilla a la gráfica
plt.savefig("grafica.png") # guarda la gráfica en un archivo
plt.show() # quita el encabezado de la gráfica
```

```python
x = np.linspace(0,40,100)
y = np.sin(x)
plt.plot(x,y)
plt.title("$F(x) = Sen(x)$")
plt.xlabel("$x$")
plt.ylabel("$Sen(x)$")
plt.xlim(0, 30) #dominio
plt.ylim(-1,1) #rango

plt.grid()
plt.show()
```

```python
x = np.linspace(0,10,100)
y =  np.sin(x)
y2 = np.cos(x)


plt.plot(x,y)
plt.plot(x,y2)
plt.title("$Sen(x), Cos(x)$")
plt.xlabel("x")
plt.legend(["$Sen(x)$","$Cos(x)$"])

plt.grid()
```

```python
#Subplots

from matplotlib import legend
plt.subplot( 2, 1, 1) #filas, #columnas,
plt.plot(x,y, label="$Sen(x)$")
plt.title("$Sen(x)$")
plt.xlabel("$x$")
plt.legend()
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(x,y2, color = "red", label="$Cos(x)$")
plt.title("$Cos(x)$")
plt.xlabel("$x$")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo"]
ventas = [10, 15, 12, 20, 25]

plt.plot(meses, ventas, marker="o")

plt.title("Ventas por mes")
plt.xlabel("Mes")
plt.ylabel("Ventas")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

actividades = ["Estudiar", "Dormir", "Comer", "Ejercicio", "Ocio"]
horas = [6, 8, 2, 2, 6]

plt.pie(horas, labels=actividades, autopct="%1.1f%%")

plt.title("Distribución de tiempo al día")

plt.show()
```
